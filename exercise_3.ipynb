#print 1 to n odd numbers

```python
n=int(input())
for i in range(1,n,2):
  print(i)
```

#print multipication of the given number

```python
n=int(input())
for i in range(1,n):
  print(n,"*",i,"=",n*i)
```

#print the number is prime or not

```python
n=int(input())
for i in range(1,n):
  if(n%i==0):
    print("the number is not prime")
    break
  else:
    print("prime")
```

#print all the prime numbers between the given numbeer

```python
a=int(input())
b=int(input())
for i in range(a,b+1):
  for j in range(2,i):
    if(i%j==0):
      break
  else:
      print(i,end=" ")
```

#print the sum of digits in a given number

```python
n=int(input())
sum=0
while(n!=0):
  rem=n%10
  sum=sum+rem
  n=n//10
print(sum)
```
